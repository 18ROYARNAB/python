```python
class Student():
    # attributes
    name=""
    age=0
    gender=""

s1=Student()
print(s1)
s1.name="arnab"
s1.age=19
s1.gender="male"
print(f"name is : {s1.name}")
print(f"age is : {s1.age}")
print(f"gender is : {s1.gender}")
```

```python
class Student():
    # attributes
    name=""
    age=0
    gender=""

    def display(self):
        print(f"My name : {self.name}, Age is : {self.age}, Gender is : {self.gender}")

s1=Student()
print(s1)
s1.name="arnab"
s1.age=19
s1.gender="male"
s1.display()
```

```python
class Student():
    # attributes
    name=""
    age=0
    gender=""

    def set_info(self):
        self.name=input("enter your name : ")
        self.age=input("enter your age : ")
        self.gender=input("enter your Gender : ")
    def display(self):
        print(f"My name : {self.name}, Age is : {self.age}, Gender is : {self.gender}")

s1=Student()
print(s1)
s1.set_info()
s1.display()
```

```python
class Student():
    # attributes
    name=""
    age=0
    gender=""
    # Methods Function inside classes are called method
    def set_info(self,name:str,age:int,gender:str): # name : str means name shall be given string type these are called Annotations
        self.name=name
        self.age=age
        self.gender=gender

    def display(self):
        print(f"My name : {self.name}, Age is : {self.age}, Gender is : {self.gender}")

s1=Student()
print(s1)
s1.set_info("arnab",19,"Male")
s1.display()
```
